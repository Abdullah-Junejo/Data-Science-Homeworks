#### Question 1: If the Monthly Income of Males is greater than Females
#### H_0 -> Monthly income of Males <= Monthly income of Females
#### H_1 -> Monthly income of Males > Monthly income of Females

```python
#performing a p-test on the data of monthly income for male and female(from gender coloumn)

SalaryMale = df[(df['Gender'] == 'Male')]['MonthlyIncome']
SalaryFemale = df[(df['Gender'] == 'Female')]['MonthlyIncome']
stats.ttest_ind(SalaryMale, SalaryFemale, alternative='greater')
```

#### Since the p-value is greater than 0.05 we accept the null hypothesis and reject the alternate hypothesis claiming that monthly income of males is less than equal to monthly income of females.

#### Question 2: If the WorkLifeBalance of Males is less than Females

#### H_0 -> Work-Life Balance of Males >= Work-Life Balance of Females
#### H_1 -> WorkLifeBalance of Males < WorkLifeBalance of Females 

```python
##performing a p-test on the data of Work-life balance for male and female(from gender coloumn)

MaleWorkBal = df[(df['Gender'] == 'Male')]['WorkLifeBalance']
FemaleWorkBal = df[(df['Gender'] == 'Female')]['WorkLifeBalance']
stats.ttest_ind(MaleWorkBal, FemaleWorkBal, alternative='less')
```

#### Since the p-value is greater than 0.05 we accept the null hypothesis and reject the alternate hypothesis claiming that the WorkLifeBalance of Males is greater than or equal to the WorkLifeBalance of females

#### Question 3: If the YearsAtCompany of Single is less than Married

#### Question 5: If the MonthlyIncome of Manager is greater than Laboratory Technician (Hint: Use JobRole to find Manager and Laboratory Technician)

#### Question 6: Is there any association between the Department and Gender?
#### H_0 -> There is no association between the Department and Gender
#### H_1 -> There is an association between the Department and Gender

```python
#Using chi square test to check if there is any relation between Department and Gender

Department = df['Department']
Gender = df['Gender']
chisqt = pd.crosstab(Department,Gender, margins=True)
value = np.array([chisqt.iloc[0][0:].values, chisqt.iloc[1][0:].values])
chi2, p, dof, ex = stats.chi2_contingency(value)
print(p)
```

#### Since the p-value is greater than 0.05 we accept the null hypothesis and reject the alternate hypothesis claiming that there is no association between the Department and Gender.
