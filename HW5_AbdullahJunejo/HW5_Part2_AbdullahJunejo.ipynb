# Analysis 1

```python
#Plot to check the relation between Department and Hourly_Rate.

sns.catplot(x="department", y="hourly_rate", kind="box", data=dataframe1)
plt.xticks(rotation=30)
```

# Analysis 2

```python
sns.heatmap(dataframe1[['performance_rating', 'age', 'monthly_income']].corr(), annot=True)
```

### The only decent yet moderate corelating is between age and monthly income. This means as employees grow older, they have  more experience and contributions which lead them to having more monthly inicome. An interesting observation would be negative corelation of performance rating with income. This means less  performing individuals are earning more than better performing ones.
