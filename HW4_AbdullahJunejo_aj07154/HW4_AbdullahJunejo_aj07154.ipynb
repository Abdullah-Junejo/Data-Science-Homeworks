```python
import pandas as pd
import numpy as np
from scipy import stats
df = pd.read_csv('EmployeeAttrition.csv')
df
```

#### Q1: The Monthly Income of Males is greater than Females?
##### Null Hypothesis: No It is not
##### Accept Hypothesis: Yes it is.

```python
Sal1, Sal2 = df[(df['Gender'] == 'Male')]['MonthlyIncome'], df[(df['Gender'] == 'Female')]['MonthlyIncome'] #Using t-test
stats.ttest_ind(Sal1, Sal2, alternative='greater')
```

#### we see that the p-value is > 0.05 making the result statistically insignificant. Therefore, we accept the null hypothesis which claims that Income of male is not greater than females.

#### Q2: The Work/Life balance of Males is less than Females
##### Null: No its not.
##### Alternate: Yes it is.

```python
#Using t-test. Bal1 is for males and Bal2 is for females.

Bal1, Bal2 = df[(df['Gender'] == 'Male')]['WorkLifeBalance'], df[(df['Gender'] == 'Female')]['WorkLifeBalance']
stats.ttest_ind(Bal1, Bal2, alternative='less')
```

#### we see that the the p-value > 0.05 making the result statistically insignificant. Therefore, we accept the null hypothesis which claims that the Work/Life Balance of Males >= Work/Life Balance of Females.

#### Q3: The 'Years at company' of Single < Married

##### Null: No its not.
##### Alternate: Yes it is.

```python
#Using t-test
married,single = df[(df['MaritalStatus'] == 'Married')]['YearsAtCompany'], df[(df['MaritalStatus'] == 'Single')]['YearsAtCompany']
stats.ttest_ind(single, married, alternative='less')
```

#### Q4: The 'Environmental Satisfaction' of 'Attrition=Yes' < 'Attrition=No'

#### Q6: Is there any association between the Department and Gender?
##### Null: No there is no association.
##### Alternate: Yes there is association.

```python
#We will be using chi square test as Department and Gender a re both categorical data.

D,G = df['Department'], df['Gender']
cont_table = pd.crosstab(D,G, margins=True) #Contingency Table
valuess = np.array([cont_table.iloc[0][0:].values, cont_table.iloc[1][0:].values])
chi2_statistic, p_value, degrees_of_freedom, expected_values = stats.chi2_contingency(valuess)
print(p_value)
```

#### We used chisquare test as both gender and department were categorical. The results of test indicated a p-value > 0.05. Therefore, We will be accepting the null hypothesis. So there is no association between gender and department. This suggests that there was no gender bias during selection.
