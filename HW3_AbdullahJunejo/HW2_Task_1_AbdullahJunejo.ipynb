### Initialization

```python
import pandas as pd
import numpy as np
```

```python
DF = pd.read_csv('loan_small.csv')
DF
```

### Writing a small block to check for duplicates and remove if they exist or ever exist in future:

```python
#Checking for duplicate rows
if (DF.duplicated().sum()):
    #Remove duplicate rows
    DF = DF.drop_duplicates()
#Checking for duplicate columns
if (DF.columns.duplicated().sum()):
    #Remove duplicate columns
    DF = DF.loc[:, ~DF.columns.duplicated()]
DF
```

### Capitalizing Columns

```python
DF.columns = [column.capitalize() for column in DF.columns] #Capitalizing Columns
DF
```

### Converting Relational Data into Numerical/Ordinal Data (Employee Length)

```python
#Closest Reliable/accurate estimate to 10+ Years is 11. That way will not be exagerrating any employee, but we will still might be understating.
#I believe Understating is better than exageratting. < 1 Year is simply 0 as there is no integer less than that.
DF['Emp_length'] = DF['Emp_length'].replace(['10+ years'], '11')
DF['Emp_length'] = DF['Emp_length'].replace(['< 1 year'], '0')
DF['Emp_length'] = DF['Emp_length'].str.extract('(\d+)').astype(int) #Converting Str data into numeric integer data.
```

```python
DF['Emp_length']
```

### Changing Loan Status to Binary

```python
#Following the suggestion stated in the quesiton, I decided to keep the answers in binary.
#In otherwords replaced str with bool for saving space.
DF['Loan_status'] = DF['Loan_status'].replace(['Fully Paid'], True) #Considered Good Status
DF['Loan_status'] = DF['Loan_status'].replace(['Charged Off'], False) #Considered Bad as Google States: "A charge-off means a lender or creditor has written the account off as a loss, and the account is closed to future charges. "
DF['Loan_status'] = DF['Loan_status'].replace(['Current'], True) #Considered Good, Making Payments on Time.
DF['Loan_status'] = DF['Loan_status'].replace(['Default'], False) #Considered Bad Naturally
DF['Loan_status'] = DF['Loan_status'].replace(['Late (31-120 days)'], False) #Considered Bad Naturally (2)
DF['Loan_status']
```

### Correcting Label 'Months'

```python
DF['Term'] = DF['Term'].str.replace('months', '') #Excluding 'month' from entries so that data becomes numerical
DF['Term'] = DF['Term'].astype(int) #Setting int type
```

```python
DF=DF.rename(columns={'Term' : 'Term_(Months)'})
DF
```

### Dealing with Null Values:

```python
DF.isnull().sum()
```

```python
#Remove rows with null in the following columns: Employee Title, Employee length, and Last_payment_d
#It is not accurate to fill in Employee Title, Length with a mode/median Their absence does not signify a zero point value. We will be making a bold assumption if we do so. Therefore, these should be dropped as they are anomalies in our dataframe. w 
DF = DF.dropna(subset=['Emp_title', 'Emp_length', 'Last_pymnt_d'])
#The Reason I chose to drop Last Payment D is it only had one null entry, so i decided to look closely into it.
#Turns out it had many irregularities/anamolies/zero in many entries. That very row this means was some sort of dummy/perhaps test data. Therefore, it should be remove as it does not make any sense at all.
DF.isnull().sum()
```

```python
#droping the Mths_since_last_delinq and Mths_since_last_record coloumns as most of the values are null.
#This means that since most accountss have never gone into delinquincy
#Most accounts don't have last record
#We can't extrapolate/fill in with mode/median values for the sake of accuracy/relliability of our data.

DF = DF.drop(['Mths_since_last_delinq', 'Mths_since_last_record'], axis=1)
DF.isnull().sum()
```
